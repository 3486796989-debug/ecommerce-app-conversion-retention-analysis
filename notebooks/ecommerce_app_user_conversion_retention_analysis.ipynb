# E-commerce App User Conversion and Retention Analysis

Portfolio-style notebook using pandas, NumPy, and matplotlib. The full raw-data processing pipeline lives in `scripts/01_run_pandas_analysis.py`; this notebook reads the processed outputs and walks through the business findings.


```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

DATA_DIR = Path('../data/processed')
monthly = pd.read_csv(DATA_DIR / 'kpi_monthly_summary.csv')
retention = pd.read_csv(DATA_DIR / 'retention_month_to_month.csv')
brands = pd.read_csv(DATA_DIR / 'top_brands.csv')
prices = pd.read_csv(DATA_DIR / 'price_bucket_analysis.csv')
monthly.head()
```

## Monthly Funnel KPIs


```python
monthly[['month', 'active_users', 'cart_users', 'buyers', 'active_user_purchase_rate', 'revenue']]
```

```python
fig, ax1 = plt.subplots(figsize=(9, 4))
ax1.bar(monthly['month'], monthly['revenue'], color='#2563EB', alpha=.85)
ax1.set_ylabel('Revenue')
ax2 = ax1.twinx()
ax2.plot(monthly['month'], monthly['active_user_purchase_rate'], color='#0F766E', marker='o')
ax2.set_ylabel('Active user purchase rate')
plt.title('Monthly Revenue and Purchase Rate')
plt.show()
```
